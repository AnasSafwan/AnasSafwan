```python
#Preprocessing data
text = re.sub(r'\[[0-9]*\]','', para)
#text = re.sub(r'\s+','',text) no need!
text = text.lower()
#text = re.sub(r'\d','',text) no need!
#text = re.sub(r'\s+','',text) no need anymore!
```

```python
#Preparing data

sentences = nltk.sent_tokenize(text)
sentences = [nltk.word_tokenize(sentence) for sentence in sentences]

#print(text)
```

```python
for i in range (len(sentences)):
  sentences[i] = [word for word in sentences[i] if word not in stopwords.words('english')]

#print(text)
```

```python
#Training the word2vec model

model = Word2Vec(sentences, min_count=1)

# 'vocab' attribute was removed in Gensim 4.0.0. Using 'index_to_key' instead.
words = model.wv.index_to_key
```

```python
#Finding word vectors
vector = model.wv['ai']

#Most similar words
similar = model.wv.most_similar('ai')
```
